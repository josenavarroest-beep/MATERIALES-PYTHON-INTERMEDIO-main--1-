### Ejercicio 1 · Factura con descuento e IVA ★★

Pida el nombre de un producto, la cantidad y el precio unitario. Calcule:
- Subtotal = cantidad × precio
- Descuento del 10 % sobre el subtotal
- IVA del 15 % sobre (subtotal − descuento)
- Total

Muestre la factura con los valores **alineados a la derecha** en 10 espacios y 2 decimales:
```
Producto: Teclado
Subtotal:      90.00
Descuento:      9.00
IVA:           12.15
TOTAL:         93.15
```

````python
# Escriba su solución aquí


# Solicitar los datos del producto
producto = input("Ingrese el nombre del producto: ")
cantidad = int(input("Ingrese la cantidad: "))
precio = float(input("Ingrese el precio unitario: "))

# Calcular subtotal, descuento, IVA y total
subtotal = cantidad * precio
descuento = subtotal * 0.10
iva = (subtotal - descuento) * 0.15
total = subtotal - descuento + iva

# Mostrar la factura
print("\n========== FACTURA ==========")
print(f"Producto: {producto}")
print(f"Subtotal:  {subtotal:>10.2f}")
print(f"Descuento: {descuento:>10.2f}")
print(f"IVA:       {iva:>10.2f}")
print(f"TOTAL:     {total:>10.2f}")
print("=============================")
````

### Ejercicio 12 · Ahorro con interés compuesto ★★

Pida el capital inicial, la tasa anual (%) y los años. Calcule el monto final con interés compuesto, $M = C(1 + t)^n$, y la ganancia. Compárelo con el interés simple ($C \cdot t \cdot n$) y muestre la diferencia, todo con `:,.2f`.

````python
# Escriba su solución aquí

# Solicitar los datos al usuario
capital = float(input("Ingrese el capital inicial: $"))
tasa = float(input("Ingrese la tasa anual (%): "))
anios = int(input("Ingrese el número de años: "))

# Convertir la tasa porcentual a decimal
tasa_decimal = tasa / 100

# Calcular el monto con interés compuesto
monto_compuesto = capital * (1 + tasa_decimal) ** anios

# Calcular la ganancia con interés compuesto
ganancia = monto_compuesto - capital

# Calcular el interés simple y su monto final
interes_simple = capital * tasa_decimal * anios
monto_simple = capital + interes_simple

# Calcular la diferencia entre los montos
diferencia = monto_compuesto - monto_simple

# Mostrar los resultados
print("\n--- RESULTADOS DEL AHORRO ---")
print(f"Monto final con interés compuesto: ${monto_compuesto:,.2f}")
print(f"Ganancia con interés compuesto: ${ganancia:,.2f}")
print(f"Monto final con interés simple: ${monto_simple:,.2f}")
print(f"Diferencia entre ambos montos: ${diferencia:,.2f}")
````

### Ejercicio 17 · Usuario y correo institucional ★★

Con `nombre = "María José Torres Ríos"` genere:
1. El **usuario**: la inicial del primer nombre + el primer apellido, en minúsculas y sin tildes → `mtorres`.
2. El **correo**: `usuario@tecazuay.edu.ec`.
3. Las **iniciales** en mayúsculas → `MJTR`.

💡 Use `split()`, índices, `lower()` y `replace()` para quitar las tildes.

````python
nombre = "María José Torres Ríos"
# Escriba su solución aquí

# Separar el nombre en palabras
partes = nombre.split()

# Obtener el primer nombre y el primer apellido
primer_nombre = partes[0]
primer_apellido = partes[2]

# Quitar las tildes y convertir a minúsculas
primer_nombre = primer_nombre.lower().replace("á", "a").replace("é", "e").replace("í", "i").replace("ó", "o").replace("ú", "u")
primer_apellido = primer_apellido.lower().replace("á", "a").replace("é", "e").replace("í", "i").replace("ó", "o").replace("ú", "u")

# Generar el usuario
usuario = primer_nombre[0] + primer_apellido

# Generar el correo institucional
correo = usuario + "@tecazuay.edu.ec"

# Generar las iniciales en mayúsculas
iniciales = "".join(palabra[0] for palabra in partes).upper()

# Mostrar los resultados
print("--- DATOS INSTITUCIONALES ---")
print(f"Nombre completo: {nombre}")
print(f"Usuario: {usuario}")
print(f"Correo: {correo}")
print(f"Iniciales: {iniciales}")
````

### Ejercicio 22 · Inventario con diccionario ★★

Con `stock = {"laptop": 12, "mouse": 45, "teclado": 30, "monitor": 8}`:
1. Se vendieron 5 mouses: actualice el stock.
2. Llegó un producto nuevo: agregue `"webcam": 15` con `update`.
3. Se dejó de vender el monitor: elimínelo con `pop` y muestre cuántas unidades tenía.
4. Consulte `"impresora"` con `get` para que muestre `0` si no existe.
5. Muestre el total de unidades, los productos en orden alfabético y el producto con más unidades (`max(stock, key=stock.get)`).

````python
stock = {"laptop": 12, "mouse": 45, "teclado": 30, "monitor": 8}
# Escriba su solución aquí

# Inventario inicial
stock = {
    "laptop": 12,
    "mouse": 45,
    "teclado": 30,
    "monitor": 8
}

# 1. Actualizar el stock después de vender 5 mouses
stock["mouse"] -= 5

# 2. Agregar un producto nuevo utilizando update()
stock.update({"webcam": 15})

# 3. Eliminar el monitor y guardar sus unidades
unidades_monitor = stock.pop("monitor")

# 4. Consultar la impresora utilizando get()
impresora = stock.get("impresora", 0)

# 5. Calcular el total de unidades
total_unidades = sum(stock.values())

# Ordenar los productos alfabéticamente
productos_ordenados = sorted(stock.keys())

# Encontrar el producto con más unidades
producto_mayor = max(stock, key=stock.get)

# Mostrar los resultados
print("--- INVENTARIO ACTUALIZADO ---")
print(f"Stock final: {stock}")
print(f"Unidades que tenía el monitor: {unidades_monitor}")
print(f"Stock de impresoras: {impresora}")
print(f"Total de unidades: {total_unidades}")
print(f"Productos en orden alfabético: {productos_ordenados}")
print(f"Producto con más unidades: {producto_mayor}")
print(f"Cantidad disponible: {stock[producto_mayor]}")
````

### Ejercicio 24 · Operaciones con conjuntos ★★

Estudiantes inscritos en tres talleres:
```python
python_t = {"Ana", "Luis", "María", "Pedro", "Sofía"}
excel_t  = {"Luis", "Carla", "Pedro", "Diego"}
ia_t     = {"Ana", "Pedro", "Diego"}
```
Muestre (ordenado con `sorted`):
1. Quiénes están en Python **y** en Excel.
2. Quiénes están en los tres talleres.
3. Quiénes están en Python pero **no** en IA.
4. Quiénes están en Python **o** en Excel, pero no en ambos (`^`).
5. Si todos los de IA están también en Python (`issubset`).
6. El total de estudiantes distintos.

````python
python_t = {"Ana", "Luis", "María", "Pedro", "Sofía"}
excel_t = {"Luis", "Carla", "Pedro", "Diego"}
ia_t = {"Ana", "Pedro", "Diego"}
# Escriba su solución aquí

# 1. Estudiantes que están en Python y Excel
python_y_excel = sorted(python_t & excel_t)

# 2. Estudiantes que están en los tres talleres
tres_talleres = sorted(python_t & excel_t & ia_t)

# 3. Estudiantes que están en Python pero no en IA
python_no_ia = sorted(python_t - ia_t)

# 4. Estudiantes que están en Python o Excel, pero no en ambos
python_o_excel = sorted(python_t ^ excel_t)

# 5. Comprobar si todos los estudiantes de IA están en Python
ia_en_python = ia_t.issubset(python_t)

# 6. Contar el total de estudiantes distintos
total_estudiantes = len(python_t | excel_t | ia_t)

# Mostrar los resultados
print("--- RESULTADOS DE LOS TALLERES ---")
print(f"1. Python y Excel: {python_y_excel}")
print(f"2. En los tres talleres: {tres_talleres}")
print(f"3. Python pero no IA: {python_no_ia}")
print(f"4. Python o Excel, pero no ambos: {python_o_excel}")
print(f"5. ¿Todos los de IA están en Python?: {ia_en_python}")
print(f"6. Total de estudiantes distintos: {total_estudiantes}")
````

Ejecute y explique en comentarios cada resultado:

```python
original = {"nombre": "Ana", "saldo": 100}
alias = original
copia = original.copy()
alias["saldo"] = 500
copia["saldo"] = 999
```
1. Muestre `original`, `alias` y `copia`.
2. Compruebe con `is` cuáles son el mismo objeto.
3. Compare con `id()` las direcciones de memoria.

### Ejercicio 30 · Ficha de un socio a partir de texto ★★

Un archivo antiguo guarda cada socio en una sola línea:
`linea = "ana pérez|0104567896|ahorros=1520.75|plazo fijo=5000"`
1. Separe la línea por `"|"`.
2. Cree un diccionario con `nombre` (formato título), `cedula` y `cuentas`, que a su vez es un diccionario `{tipo: saldo}` (separe cada cuenta por `"="` y convierta el saldo a `float`).
3. Muestre la ficha con este formato:
```
Socio: Ana Pérez (0104567896)
Tipos de cuenta: ahorros, plazo fijo
Saldo total: $6,520.75
```

````python
linea = "ana pérez|0104567896|ahorros=1520.75|plazo fijo=5000"
# Escriba su solución aquí

# 1. Separar la línea usando el símbolo |
partes = linea.split("|")

# 2. Obtener los datos principales
nombre = partes[0].title()
cedula = partes[1]

# Crear un diccionario vacío para las cuentas
cuentas = {}

# Recorrer las cuentas y separar el tipo de su saldo
for cuenta in partes[2:]:
    tipo, saldo = cuenta.split("=")
    cuentas[tipo] = float(saldo)

# Crear el diccionario del socio
socio = {
    "nombre": nombre,
    "cedula": cedula,
    "cuentas": cuentas
}

# 3. Mostrar la ficha del socio
print("Socio:", socio["nombre"], f"({socio['cedula']})")

# Mostrar los tipos de cuenta
tipos_cuenta = ", ".join(socio["cuentas"].keys())
print("Tipos de cuenta:", tipos_cuenta)

# Calcular el saldo total de todas las cuentas
saldo_total = sum(socio["cuentas"].values())
print(f"Saldo total: ${saldo_total:,.2f}")

# Explicación:
# split("|") separa los datos de la línea.
# title() convierte el nombre a formato título.
# split("=") separa el tipo de cuenta y su saldo.
# float() convierte el saldo de texto a número decimal.
# sum() suma los saldos de todas las cuentas.
# join() une los tipos de cuenta para mostrarlos.
````
